Exercício 5 — Transformação e resumo
Enunciado: Converta a data, crie tempo_dias e calcule tempo médio por categoria em ordem decrescente. Interprete a maior média.

```python
df["data_abertura"] = pd.to_datetime(df["data_abertura"])
df["tempo_dias"] = df["tempo_resolucao_horas"] / 24
media = df.groupby("categoria")["tempo_resolucao_horas"].mean().sort_values(ascending=False)
media
```

Atividade final — Desafio "Painel de atenção do suporte"
SLA: Crítica = 4 h; Alta = 8 h; Média = 24 h; Baixa = 48 h.
- Inspecionar shape, info() e ao menos uma estatística.
- Selecionar as colunas relevantes e filtrar chamados resolvidos.
- Criar limite_sla_horas e dentro_sla.
- Calcular quantidade e percentual dentro/fora do SLA.
- Calcular tempo médio por categoria.
- Concluir em 3–5 linhas com duas evidências numéricas e uma recomendação.

```python
print(df.shape)
df.info()
df["tempo_resolucao_horas"].mean()
```

```python
colunas = ["id_chamado", "categoria", "prioridade", "status",
           "tempo_resolucao_horas", "satisfacao", "reaberto"]
analise = df[colunas].copy()
resolvidos = analise[analise["status"] == "Resolvido"].copy()
resolvidos
```

```python
limites = {"Crítica": 4, "Alta": 8, "Média": 24, "Baixa": 48}
resolvidos["limite_sla_horas"] = resolvidos["prioridade"].map(limites)
resolvidos["dentro_sla"] = resolvidos["tempo_resolucao_horas"] <= resolvidos["limite_sla_horas"]
resolvidos
```

```python
contagem = resolvidos["dentro_sla"].value_counts()
percentual = resolvidos["dentro_sla"].value_counts(normalize=True).mul(100).round(1)
print(contagem)
print(percentual)
```

```python
media_categoria = resolvidos.groupby("categoria")["tempo_resolucao_horas"].mean().sort_values(ascending=False).round(2)
media_categoria
```

Conclusão: Dos 16 chamados resolvidos, 11 ficaram dentro do SLA (68,8%) e 5 ficaram fora (31,2%). A categoria Sistema lento tem a maior média de tempo de resolução (47,75 h), puxada pelo chamado de 120 h. Recomendação: priorizar a investigação da categoria Sistema lento, já que ela concentra os maiores tempos de resolução e afeta diretamente o cumprimento do SLA. Os chamados ainda não resolvidos (status diferente de Resolvido) não entram no tempo final porque ainda não têm um tempo de resolução definido — incluí-los distorceria a média.
